```python
fig = px.choropleth(data_frame=final_df, locations='state_code', locationmode='USA-states', 
                    color='disparity_value_num', scope='usa', color_continuous_scale='Reds', animation_frame='year', animation_group='state_code',
                    labels={'disparity_value_num': 'Smoking Disparity Between Severe Distress and No Distress'}, 
                    title='Smoke Disparities of Heavy Stress Vs No Stress Levels From 2011-2023') 

fig.show()
```

Summary: This choropleth shows the smoking disparity between 'High Distress' and 'No Distress' mental groups accross all states from 2011-2023. The year slider allows one to compare the disparity rates between years. We see large disparities between the years of 2013-2015 with a slight drop off after that. This then reaches the largest disparity spread in the country in 2019 with a huge drop off in disparity during COVID/quarentine in 2020. One could begin to therorize if the time off from work during quarentine could have led to the disparity between these two groups decreasing significantly.

```python

prev_df = df[['year', 'focus_prevalence_num']]

```

```python
fig2 = px.histogram(data_frame=prev_df, x='focus_prevalence_num', facet_col='year', 
                    nbins=20, labels={'focus_prevalence_num': 'Smoker %'}, facet_col_wrap=7,
                    facet_col_spacing=.03, facet_row_spacing=.05,
                    title='Percentage of all Smokers by Year') 
fig2.update_layout(bargap=.05)
fig2.update_xaxes(tickangle=45) 
fig2.show()
```

Summary: These graphs show the distribution of the percentages of smokers across all demographics from 2011-2023. On average we can see that 15-19.9% of all demographics were smokers. The percentage bins larger than 15-19.9% seem to be significantly lower. We begin to see change in the percentages from 2021-2023 where we see a large increase in the percentages of 5-14.99%, and a decrease in the 15-19.9%. This would mean that within the past few years we have seen an overall descrease in the amount of smokers accross all of the focus groups.
